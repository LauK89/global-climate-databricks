##Catálogo y esquemas (Unity Catalog)

Un solo catálogo para todo el proyecto. No hace falta separar por entorno (dev/prod) en un portfolio personal — eso tiene sentido en catálogos de empresa con varios equipos, pero aquí solo añadiría ruido.

```python
%sql
CREATE CATALOG IF NOT EXISTS global_climate
COMMENT 'Portfolio: análisis de cambio climático y fenómenos extremos';

CREATE SCHEMA IF NOT EXISTS global_climate.bronze
COMMENT 'Datos crudos tal cual llegan de las fuentes (NASA, NOAA, AEMET)';

CREATE SCHEMA IF NOT EXISTS global_climate.silver
COMMENT 'Datos limpios, tipados y con feature engineering';

CREATE SCHEMA IF NOT EXISTS global_climate.gold
COMMENT 'Agregados listos para BI, ML y el agente';

CREATE SCHEMA IF NOT EXISTS global_climate.ml
COMMENT 'Feature tables, experimentos y modelos registrados';
```

```python
2. Volumes — un volumen por fuente, no uno gigante

Dentro de bronze, crea un volumen gestionado por cada fuente de datos, así el linaje queda clarísimo y puedes dar permisos distintos si algún día quieres.
Y dentro de cada volumen, estructura por fecha de ingesta (te da trazabilidad de cuándo llegó cada versión del dato, importante porque GISTEMP se actualiza mensualmente):%md
```

```python
%sql
CREATE VOLUME IF NOT EXISTS global_climate.bronze.nasa_gistemp;
CREATE VOLUME IF NOT EXISTS global_climate.bronze.noaa_ibtracs;
CREATE VOLUME IF NOT EXISTS global_climate.bronze.aemet_dana;
```
